```python
# ============================================================
# SECTION 1 : IMPORT LIBRARIES
# ============================================================

import pandas as pd
import zipfile
import os
import glob
```

```python
# ============================================================
# SECTION 2 : DEFINE FILE PATHS
# ============================================================

zip_file = "Datasets-shoping-app-main.zip"
extract_folder = "shopping_dataset"
```

```python
# ============================================================
# SECTION 3 : EXTRACT ZIP FILE
# ============================================================

with zipfile.ZipFile(zip_file, 'r') as zip_ref:
    zip_ref.extractall(extract_folder)

print("ZIP file extracted successfully!")
```

```python
# ============================================================
# SECTION 4 : VIEW EXTRACTED FILES
# ============================================================

os.listdir(extract_folder)
```

```python
# ============================================================
# SECTION 5 : FIND ALL CSV FILES
# ============================================================

csv_files = glob.glob(
    extract_folder + "/**/*.csv",
    recursive=True
)

print(csv_files)
```

```python
# ============================================================
# SECTION 6 : LOAD ALL CSV FILES
# ============================================================

dataframes = []

for file in csv_files:
    df = pd.read_csv(file)
    dataframes.append(df)

print("All CSV files loaded successfully!")
```

```python
# ============================================================
# SECTION 7 : COMBINE ALL DATAFRAMES
# ============================================================

combined_df = pd.concat(
    dataframes,
    ignore_index=True
)

print("All CSV files combined successfully!")
```

```python
# ============================================================
# SECTION 8 : SAVE COMBINED DATASET
# ============================================================

combined_df.to_csv(
    "combine_shopping_dataset.csv",
    index=False
)

print("Combined dataset saved!")
```

```python
# ============================================================
# SECTION 16 : FILL MISSING VALUES
# ============================================================

combined_df.fillna(
    "Unknown",
    inplace=True
)

combined_df.isnull().sum()
```

```python
# ============================================================
# SECTION 17 : DROP NULL VALUES
# ============================================================

combined_df.dropna(inplace=True)

combined_df.isnull().sum()
```

```python
# ============================================================
# SECTION 19 : FILTER HIGH RATED PRODUCTS
# ============================================================

high_rating = combined_df[
    combined_df["rating"] > 4
]

high_rating.head()
```

```python
# ============================================================
# SECTION 20 : REMOVE DUPLICATES
# ============================================================

combined_df.drop_duplicates(
    inplace=True
)

print("Duplicate rows removed!")
```

```python
# ============================================================
# SECTION 21 : CREATE QUANTITY COLUMN
# ============================================================

combined_df["quantity"] = 2
```

```python
# ============================================================
# SECTION 22 : CREATE TOTAL_AMOUNT COLUMN
# ============================================================

combined_df["total_amount"] = (
    combined_df["final_price"] *
    combined_df["quantity"]
)
```

```python
# ============================================================
# SECTION 23 : SAVE CLEANED DATASET
# ============================================================

combined_df.to_csv(
    "cleaned_shopping_dataset.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
